```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
#sets x and y as the specified array
x = [1,2,3,4,5]
y = [5,4,5,3,2]

plt.plot(x,y, "g^") #plots x vs y, with x as the x axis. The icons are green triangles
plt.xlabel("x")#labels the x axis as x
plt.ylabel("y")#labels the y axis as y
plt.title("x vs. y")#sets the title as x vs. y
plt.show()#displays graph
```
